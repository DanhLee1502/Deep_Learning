### **Bài 6. Phép toán và nhân ma trận**
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

Shape của a @ b.T: (2, 2). Lấy ma trận (2, 3) nhân với ma trận (3, 2) sẽ tạo ra ma trận (2, 2).

### **Bài 7. Các hàm thống kê trên Tensor**

### **Bài 8. Autograd cơ bản**
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

Tính thủ công:
- Hàm số: $y = x^2$
- Đạo hàm: $y' = 2x$
- Tại $x = 3.6$, giá trị đạo hàm là $y' = 2 \times 3.6 = 7.2$.
- **Nhận xét:** Kết quả tính thủ công khớp hoàn toàn với giá trị x.grad.

Code chạy thử cho hàm $y = 3x^2 + 2x + 1$:

```python
import torch
x = torch.tensor(3.6, requires_grad=True)
y = 3 * x**2 + 2 * x + 1
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

Hàm số: $y = 3x^2 + 2x + 1$

Đạo hàm: $y' = 6x + 2$

Tại $x = 3.6$:
- Giá trị $y = 3(3.6)^2 + 2(3.6) + 1 = 47.08$ (khớp với kết quả).
- Giá trị đạo hàm $y' = 6(3.6) + 2 = 23.6$ (khớp với x.grad).

### **Bài 9. Gradient Descent tìm hệ số đa thức**
Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(3000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
print("Final MSE:", mse.item())
```

So sánh hệ số:
- Hệ số học được [1.0000, 2.0000, 3.0000] gần như trùng khớp hoàn toàn với hệ số gốc [1, 2, 3] (MSE cuối ~6.6e-10). Việc đặt seed cố định và tăng số epoch lên 3000 giúp mô hình hội tụ ổn định hơn.

Giải thích các hàm:
- optimizer.zero_grad(): Xóa gradient của bước trước về 0 để tránh tình trạng cộng dồn sai lệch.
- mse.backward(): Tính đạo hàm của hàm mất mát theo tham số cần tối ưu (w).
- optimizer.step(): Cập nhật lại giá trị tham số dựa trên gradient vừa tính.

### **Bài 10. Giải hệ phương trình bằng Autograd**
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Kiểm tra nghiệm:

Các giá trị hoàn toàn thỏa mãn hệ phương trình (với sai số cực nhỏ):
- $A + B = 8.9999 \approx 9$
- $C - D = 1.0001 \approx 1$
- $A + C = 8$
- $B - D = 2$

Lý do hệ có nhiều nghiệm:

- Hệ bị phụ thuộc tuyến tính: Phương trình 4 là hệ quả của 3 phương trình đầu: $(A+B) - (A+C) + (C-D) = B - D = 2$.
- Có 4 ẩn nhưng chỉ 3 phương trình độc lập dẫn đến vô số nghiệm. Thuật toán chỉ tìm ra một nghiệm ngẫu nhiên phụ thuộc vào giá trị khởi tạo ban đầu.

### **Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:**
- Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
- Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
- Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
- Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import torch
import numpy as np

torch.manual_seed(0)
np.random.seed(0)

# Yêu cầu 1: Khởi tạo và Reshape
X = torch.arange(24)
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)

print(X_4x6.shape)
print(X_2x12.shape)
print(X_2x3x4.shape)

# Yêu cầu 2: Phép toán và Thống kê
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print(A + B)
print(A * B)
print(A / B)
print(torch.mean(A, dim=0))
print(torch.std(A, dim=0))

# Yêu cầu 3: Autograd
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
print(x.grad.item())

# Yêu cầu 4: Gradient Descent tìm hệ số
N = 100
X_np = np.random.randn(N, 1) * 5
Y_np = 2 * X_np**2 - 3 * X_np + 5

XX = np.hstack([X_np**2, X_np, np.ones_like(X_np)])
w = torch.randn(3, 1, requires_grad=True)

X_tensor = torch.tensor(XX, dtype=torch.float32)
Y_tensor = torch.tensor(Y_np, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(5000):
    optimizer.zero_grad()
    Y_pred = X_tensor @ w
    mse = torch.mean((Y_tensor - Y_pred)**2)
    mse.backward()
    optimizer.step()

print("\nHệ số học được (so với [2, -3, 5]):\n", w.detach().numpy())
print("Final MSE:", mse.item())
```

Kiểm tra thủ công (Yêu cầu 3): Đạo hàm của $y = 2x^2 + 5x + 3$ là $y' = 4x + 5$. Thay $x = 2$, ta được kết quả là 13.0, khớp chính xác với kết quả từ x.grad.

Hệ số học được (Yêu cầu 4): Nhờ việc sử dụng 5000 epoch, sai số (MSE) đã giảm xuống mức cực kỳ nhỏ ($9.28 \times 10^{-7}$). 

Các hệ số học được là [2.000019, -3.0000048, 4.998782], gần như tuyệt đối chính xác so với hệ số gốc của đa thức là [2, -3, 5].
